# Compare LLM Output to OpenSpiel

Use this notebook for a first manual comparison between a generated Python file and an OpenSpiel reference (or a manually collected reference file).


```python
from pathlib import Path
from difflib import unified_diff

ROOT = Path.cwd().resolve()
OUTPUTS = ROOT / 'outputs'

llm_code_path = OUTPUTS / 'replace_me__model.py'
reference_path = OUTPUTS / 'replace_me__openspiel_reference.py'

print('Notebook root:', ROOT)
print('LLM code path:', llm_code_path)
print('Reference path:', reference_path)
```

```python
def read_text(path: Path) -> str:
    if not path.exists():
        return f'FILE NOT FOUND: {path}'
    return path.read_text(encoding='utf-8')

llm_text = read_text(llm_code_path)
ref_text = read_text(reference_path)

print(llm_text[:4000])
print('\n' + '=' * 80 + '\n')
print(ref_text[:4000])
```

```python
llm_lines = llm_text.splitlines(keepends=True)
ref_lines = ref_text.splitlines(keepends=True)

diff = unified_diff(
    ref_lines,
    llm_lines,
    fromfile=str(reference_path.name),
    tofile=str(llm_code_path.name),
)

print(''.join(diff)[:12000])
```
